```python
import csv

# 读取 CSV 文件中的 idx 列并返回集合
def read_idx_from_csv(file_path):
    with open(file_path, mode='r') as file:
        reader = csv.reader(file)
        return {row[0].strip() for row in reader}  # 使用集合推导式

# 读取两个 CSV 文件中的 idx
csv_file_1 = 'FBDB15K_predicted_wrong_xy.csv'
csv_file_2 = 'FBDB15K_missing_idx'

idx_set_1 = read_idx_from_csv(csv_file_1)
idx_set_2 = read_idx_from_csv(csv_file_2)

# 求交集并排序
intersection = sorted(idx_set_1 & idx_set_2)  # 使用集合交集运算符 &，并排序结果

# 将交集结果保存到一个新的 CSV 文件中
with open('FBDB15K_wrong_missing.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    for idx in intersection:
        writer.writerow([idx])  # 写入交集中的 idx

# 输出交集
print(f"交集大小: {len(intersection)}")
print(f"交集内容: {intersection}")
```
